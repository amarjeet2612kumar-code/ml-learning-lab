```python
import numpy as np

# 1. Create a dataset: rows = students, columns = features
# Columns: [study_hours, attendance_percentage, marks]
data = np.array([[2,60,45],
                 [4,70,50],
                 [6,80,55],
                 [8,90,65],
                 [10,100,75]
                 ],dtype=float)
print("Dataset:\n", data)

# 2. Check the dataset structur
print("Shape of the dataset:", data.shape)
print("Dimesions of the dataset:", data.ndim)
print("Size of the dataset:", data.size)
print("Datatype of the dataset:", data.dtype)
```

```python
# 3. Select rows and columns
print("First Student row of the dataset:", data[0])
print("Study Hours",data[:,0])
print("Marks",data[:,2])
```

```python
# 4. Filter students who scored at least 70
print("High scoring students:", data[data[:,2]>=70])
```

```python
# 5. Calculate statistics
print("Maximum marks:", np.max(data[:,2]))
print("Average marks:", np.mean(data[:,2]))
print("Average Marks by column:", np.mean(data, axis=0))
```

```python
# 6. Add a new feature: study efficiency
# Marks per study hour; avoid division by zero in real datasets.
study_efficiency = data[:, 2] / data[:, 0]
print("Marks per study hour:", study_efficiency)
```

```python
# 7. Handle missing values
data_with_missing = data.copy()
data_with_missing[1, 2] = np.nan
print("Missing values by column:", np.isnan(data_with_missing).sum(axis=0))
```

```python
# Replace missing marks with the mean of the observed marks.
marks = data_with_missing[:, 2]
mean_marks = np.nanmean(marks)
marks[np.isnan(marks)] = mean_marks

print("Marks after imputation:", marks)
```

```python
# 8. Normalize study hours using min-max scaling
study_hours = data[:, 0]
normalized_hours = (
    (study_hours - np.min(study_hours))
    / (np.max(study_hours) - np.min(study_hours))
)
print("Normalized study hours:", normalized_hours)
```

```python
# 9. Separate features (X) and target (y)
X = data[:, :2]  # Study hours and attendance
y = data[:, 2]   # Marks to predict

print("Features (X):\n", X)
print("Target (y):", y)
```

```python
# 10. Add a new student's data
new_student = np.array([[5, 80]])
X_extended = np.vstack((X, new_student))

print("Updated feature shape:", X_extended.shape)
```

## Most important indexing pattern

| Code                    | Meaning                            |
| ----------------------- | ---------------------------------- |
| `data[0, :]`            | First row, all columns             |
| `data[:, 0]`            | All rows, first column             |
| `data[:, :2]`           | All rows, first two columns        |
| `data[:, 2]`            | All rows, third column             |
| `data[1:3, :]`          | Second and third rows, all columns |
| `data[data[:, 2] > 70]` | Entire rows where marks exceed 70  |

Remember: Python uses zero-based indexing, so column index `2` means the third column. In machine learning, `X` commonly holds the input features, while `y` holds the target value the model is learning to predict.

## Understand the two most important concepts
`axis=0` versus `axis=1`

Given this array:-
data = np.array([    [2, 70, 60],    [4, 80, 75],    [6, 90, 90]])

- `data.sum(axis=0)` → `[12, 240, 225]` — sums each column.
- `data.sum(axis=1)` → `[132, 159, 186]` — sums each row.



```python
import numpy as np

# Columns: Study hours, Attendance, Marks
data = np.array([
    [2.0, 70.0, 60.0],
    [4.0, np.nan, 75.0],
    [6.0, 90.0, np.nan],
    [3.0, 75.0, 65.0],
    [5.0, 85.0, 80.0]
])

print("Original dataset:\n", data)
```

```python
# 1. Identify missing values
missing_mask = np.isnan(data)
print("\nMissing value locations:\n", missing_mask)
```

```python
# 2. Count missing values
print("\nTotal missing values:", np.isnan(data).sum())
print("Missing values per column:", np.isnan(data).sum(axis=0))
```

```python
# 3. Calculate mean while ignoring missing values
column_means = np.nanmean(data, axis=0)
print("\nColumn means:", column_means)
```

```python
# 4. Replace missing values with the mean of each column
cleaned_data = data.copy()

for column in range(cleaned_data.shape[1]):
    missing_rows = np.isnan(cleaned_data[:, column])
    cleaned_data[missing_rows, column] = column_means[column]

print("\nCleaned dataset:\n", cleaned_data)
```

```python
# 5. Filter students whose marks are at least 70
high_scorers = cleaned_data[cleaned_data[:, 2] >= 70]
print("\nStudents scoring at least 70:\n", high_scorers)
```

```python
# 6. Apply two conditions
selected_students = cleaned_data[
    (cleaned_data[:, 1] >= 80) &
    (cleaned_data[:, 2] >= 75)
]
print("\nAttendance >= 80 and marks >= 75:\n", selected_students)
```

```python
# 7. Find unique marks
print("\nUnique marks:", np.unique(cleaned_data[:, 2]))
```

```python
# 8. Sort marks
print("Sorted marks:", np.sort(cleaned_data[:, 2]))
```

```python
import numpy as np

# Each row represents one student.
# Columns: Study hours, Attendance, Marks

data = np.array([
    [2, 70, 60],
    [4, 80, 75],
    [6, 90, 90],
    [3, 75, 65],
    [5, 85, 80],
    [7, 95, 95],
    [1, 60, 50],
    [8, 98, 98],
    [4, 82, 78],
    [6, 88, 87]
])

# 1. Separate input features (X) and target (y)
X = data[:, :2]  # Study hours and attendance
y = data[:, 2]   # Marks

print("Input features:\n", X)
print("Target values:\n", y)
```

```python
# 2. Create shuffled indexes reproducibly
np.random.seed(42)
indices = np.random.permutation(len(X))

print("Shuffled indexes:", indices)
```

```python
# 3. Shuffle X and y using the same indexes
X_shuffled = X[indices]
y_shuffled = y[indices]
print("Shuffled features:\n", X_shuffled)
print("Shuffled targets:\n", y_shuffled)    
```

```python
# 4. Calculate the 80% training split
split = int(len(X_shuffled) * 0.8)
print("Training split index:", split)
```

```python
# 5. Split input features
X_train = X_shuffled[:split]
X_test = X_shuffled[split:]
print("Training features:\n", X_train)
print("Testing features:\n", X_test)
```

```python
# 6. Split target values using the same split
y_train = y_shuffled[:split]
y_test = y_shuffled[split:]
print("Training targets:\n", y_train)
print("Testing targets:\n", y_test)
```

```python
# 7. Inspect the results
print("\nTraining input shape:", X_train.shape)
print("Testing input shape:", X_test.shape)

print("\nTraining target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

print("\nX_train:\n", X_train)
print("y_train:", y_train)

print("\nX_test:\n", X_test)
print("y_test:", y_test)
```

```python
import numpy as np

# Rows: students
# Columns: study hours, attendance, marks

data = np.array([
    [2.0, 70.0, 60.0],
    [4.0, 80.0, 75.0],
    [6.0, 90.0, 90.0]
])

print("Original data:\n", data)
```

```python
# 1. Broadcasting: add 10 to every element
print("\nAdd 10:\n", data + 10)
```

```python
# 2. Broadcasting: add a different value to each column
adjustment = np.array([1.0, 2.0, 3.0])
print("\nColumn-wise adjustment:\n", data + adjustment)
```

```python
# 3. Standardize each column by subtracting its mean
column_means = data.mean(axis=0)
centered_data = data - column_means
print("\nColumn means:", column_means)
print("Centered data:\n", centered_data)
```

```python
# 4. Vectorization: no explicit loop needed
study_hours = data[:, 0]
attendance = data[:, 1]
engagement_score = study_hours * attendance
print("\nEngagement scores:", engagement_score)
```

```python
# 5. Use a condition on all values
marks = data[:, 2]
pass_indicator = np.where(marks >= 70, 1, 0)
print("\nMarks:", marks)
print("Pass indicators:", pass_indicator)
```

```python
# 6. Calculate prediction errors
actual = np.array([60.0, 75.0, 90.0])
predicted = np.array([65.0, 72.0, 85.0])

errors = actual - predicted
squared_errors = errors ** 2
mse = np.mean(squared_errors)

print("\nErrors:", errors)
print("Squared errors:", squared_errors)
print("Mean squared error:", mse)
```
