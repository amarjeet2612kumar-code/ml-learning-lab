```python
import matplotlib.pyplot as plt


# -----------------------------
# Data
# -----------------------------

study_hours = [1, 2, 3, 4, 5]
marks = [45, 55, 65, 72, 85]

cities = ["Bengaluru", "Delhi", "Mumbai"]
customers = [120, 90, 110]

```

```python
# -----------------------------
# Line Plot
# -----------------------------

plt.plot(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.grid()

plt.show()
```

```python
# -----------------------------
# Scatter Plot
# -----------------------------

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
# -----------------------------
# Bar Chart
# -----------------------------

plt.bar(cities, customers)

plt.xlabel("City")
plt.ylabel("Customers")
plt.title("Customers by City")

plt.show()
```

```python
# -----------------------------
# Histogram
# -----------------------------

plt.hist(marks, bins=5)

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")

plt.show()
```

| API | Purpose |
|---|---|
| `plt.plot()` | Line plot |
| `plt.scatter()` | Scatter plot |
| `plt.bar()` | Bar chart |
| `plt.hist()` | Histogram |
| `plt.xlabel()` | X-axis label |
| `plt.ylabel()` | Y-axis label |
| `plt.title()` | Chart title |
| `plt.legend()` | Show legend |
| `plt.grid()` | Show grid |
| `plt.xlim()` | Set X-axis range |
| `plt.ylim()` | Set Y-axis range |
| `plt.show()` | Display chart |
| `plt.figure()` | Create a figure |

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.DataFrame({
    "student": ["Amar", "Rahul", "Priya", "Ankit", "Neha"],
    "study_hours": [1, 2, 3, 4, 5],
    "marks": [45, 55, 65, 72, 85],
    "attendance": [70, 75, 82, 88, 95]
})
```

```python
# -----------------------------------
# Scatter: study hours vs marks
# -----------------------------------

df.plot(
    x="study_hours",
    y="marks",
    kind="scatter"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()

plt.show()
```

```python
# -----------------------------------
# Line: study hours vs marks
# -----------------------------------

df.plot(
    x="study_hours",
    y="marks",
    kind="line"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# -----------------------------------
# Histogram: marks distribution
# -----------------------------------

df["marks"].plot(
    kind="hist",
    bins=5
)

plt.title("Marks Distribution")
plt.xlabel("Marks")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.DataFrame({
    "city": [
        "Bengaluru", "Delhi", "Bengaluru", "Mumbai",
        "Delhi", "Bengaluru", "Mumbai", "Delhi"
    ],
    "gender": [
        "Male", "Male", "Female", "Male",
        "Female", "Male", "Female", "Female"
    ],
    "study_hours": [2, 3, 4, 5, 6, 8, 4, 7],
    "attendance": [70, 75, 82, 85, 90, 95, 78, 92],
    "marks": [50, 55, 65, 72, 82, 95, 68, 88]
})
```

```python
# ------------------------------------
# 1. Numerical vs Numerical
# ------------------------------------

sns.scatterplot(
    data=df,
    x="study_hours",
    y="marks"
)
plt.title("Study Hours vs Marks")
plt.show()
```

```python
# ------------------------------------
# 2. Numerical vs Categorical
# ------------------------------------

sns.boxplot(
    data=df,
    x="city",
    y="marks"
)

plt.title("Marks by City")
plt.show()
```

```python
# ------------------------------------
# 3. Average numerical value
# ------------------------------------

sns.barplot(
    data=df,
    x="city",
    y="marks"
)

plt.title("Average Marks by City")
plt.show()
```

```python
# ------------------------------------
# 4. Categorical vs Categorical
# ------------------------------------

sns.countplot(
    data=df,
    x="city",
    hue="gender"
)

plt.title("Gender Distribution by City")
plt.show()
```

```python
# ------------------------------------
# 5. Distribution by category
# ------------------------------------

sns.histplot(
    data=df,
    x="marks",
    hue="gender",
    bins=5
)

plt.title("Marks Distribution by Gender")
plt.show()
```

```python
# ------------------------------------
# 6. Multiple numerical relationships
# ------------------------------------

sns.pairplot(
    df[
        ["study_hours", "attendance", "marks"]
    ]
)

plt.show()
```

When to use each plot
| Plot | Data type | Use when you want to know... | Example | Main library/API |
|---|---|---|---|---|
| **Line Plot** | Numerical + ordered/time | **How does something change over time/order?** | Revenue by month | `plt.plot()` |
| **Bar Plot** | Categorical + numerical | **How do categories compare?** | Average marks by city | `plt.bar()`, `sns.barplot()` |
| **Histogram** | One numerical variable | **How is the data distributed?** | Distribution of salary | `plt.hist()`, `sns.histplot()` |
| **Scatter Plot** | Two numerical variables | **Are two variables related?** | Experience vs salary | `plt.scatter()`, `sns.scatterplot()` |
| **Box Plot** | Numerical + optional category | **What is the spread and are there potential outliers?** | Salary distribution by department | `plt.boxplot()`, `sns.boxplot()` |
| **Count Plot** | One categorical variable | **How many records are in each category?** | Customers by city | `sns.countplot()` |
| **Heatmap** | Numerical matrix | **Which variables are strongly/weakly related?** | Feature correlation | `sns.heatmap()` |
| **Pair Plot** | Multiple numerical variables | **How do multiple features relate to each other?** | Age, income, spending | `sns.pairplot()` |


The easiest way to remember
| Question | Plot |
|---|---|
| **Trend over time?** | 📈 Line |
| **Compare categories?** | 📊 Bar |
| **Distribution of one number?** | 📊 Histogram |
| **Relationship between two numbers?** | 🔵 Scatter |
| **Outliers / spread?** | 📦 Box |
| **Count categories?** | 📊 Count |
| **Correlation between many features?** | 🔥 Heatmap |
| **Relationships among many numerical features?** | 🔵 Pair plot |

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "age": [20, 21, 22, 23, 24, 25, 26, 27],
    "study_hours": [2, 3, 4, 5, 6, 7, 8, 9],
    "attendance": [60, 65, 70, 75, 80, 85, 90, 95],
    "marks": [50, 55, 60, 65, 72, 78, 85, 90]
})
```

```python
corr = df.corr(numeric_only=True)

print(corr)
```

Correlation ranges from:
| Correlation | Meaning |
|---:|---|
| `+1` | Perfect positive relationship |
| `+0.7` | Strong positive relationship |
| `0` | Little/no linear relationship |
| `-0.7` | Strong negative relationship |
| `-1` | Perfect negative relationship |

```python
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True)

plt.title("Feature Correlation")
plt.show()
```

What annot=True does ?

It displays the actual correlation value inside each cell.

Heatmap vs Pairplot
| Tool | What it tells you | Best use |
|---|---|---|
| Correlation matrix | Numerical correlation values | Quantify relationships |
| Heatmap | Visual correlation matrix | Quickly identify strong relationships |
| Pairplot | Visual pairwise relationships | Explore patterns between many features |

What you should remember
For your AI/ML preparation, remember these questions rather than memorizing syntax:

| Question | Plot |
|---|---|
| How is one numerical feature distributed? | Histogram |
| Are there outliers? | Box plot |
| Are two numerical variables related? | Scatter |
| How strongly are numerical features correlated? | Heatmap |
| How do multiple numerical features relate? | Pairplot |
