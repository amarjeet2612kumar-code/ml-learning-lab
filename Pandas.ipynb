```python
import pandas as pd

# 1. Create a DataFrame
data = {
    "name": ["Amit", "Priya", "Rahul", "Neha", "Vikram"],
    "study_hours": [2, 4, 6, 3, 5],
    "attendance": [70, 80, 90, 75, 85],
    "marks": [60, 75, 90, 65, 80]
}
df = pd.DataFrame(data)

print("Complete dataset:")
print(df)
```

```python
# 2. Inspect the dataset
print("\nFirst 3 rows:")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
print(df.describe())

```

```python
# 3. Select columns
print("\nStudent names:")
print(df["name"])

print("\nNames and marks:")
print(df[["name", "marks"]])

```

```python
# 4. Select rows and values by label
print("\nFirst row using loc:")
print(df.loc[0])

print("\nMarks of first student using loc:")
print(df.loc[0, "marks"])

```

```python
# 5. Select rows and values by position
print("\nFirst row using iloc:")
print(df.iloc[0])

print("\nFirst row, third column:")
print(df.iloc[0, 2])

```

```python
# 6. Save the dataset
df.to_csv("students.csv", index=False)

print("\nDataset saved to students.csv")
```

 Pandas operations quick reference
| Operation               | Example                                | One-line explanation                                       |
| ----------------------- | -------------------------------------- | ---------------------------------------------------------- |
| Import Pandas           | `import pandas as pd`                  | Imports Pandas using the alias `pd`.                       |
| Create a DataFrame      | `pd.DataFrame(data)`                   | Creates a table with rows and columns.                     |
| Create a Series         | `pd.Series([10, 20, 30])`              | Creates a one-dimensional labeled sequence.                |
| Read CSV                | `pd.read_csv("students.csv")`          | Loads a CSV file into a DataFrame.                         |
| Save CSV                | `df.to_csv("output.csv", index=False)` | Saves a DataFrame without writing its index as a column.   |
| View first rows         | `df.head()`                            | Displays the first five rows by default.                   |
| View last rows          | `df.tail()`                            | Displays the last five rows by default.                    |
| Check dimensions        | `df.shape`                             | Returns the number of rows and columns.                    |
| Inspect column names    | `df.columns`                           | Returns all column labels.                                 |
| Inspect data types      | `df.dtypes`                            | Shows each column's data type.                             |
| Dataset summary         | `df.info()`                            | Shows column types, non-null counts, and memory usage.     |
| Statistical summary     | `df.describe()`                        | Summarizes numerical columns statistically.                |
| Select a column         | `df["marks"]`                          | Returns one column as a Series.                            |
| Select multiple columns | `df[["name", "marks"]]`                | Returns selected columns as a DataFrame.                   |
| Select by label         | `df.loc[0, "marks"]`                   | Selects a value using row and column labels.               |
| Select by position      | `df.iloc[0, 2]`                        | Selects a value using zero-based row and column positions. |


```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Amar", "Rahul", "Priya"],
    "marks": [85, 72, 91],
    "age": [25, 27, 24]
}, index=["A", "B", "C"])

print(df)
# 1. Select a row using its label
print(df.loc["B"])

# 2. Select the first row using its position
print(df.iloc[0])

# 3. Select marks for index A
print(df.loc["A", "marks"])  # 85

# 4. Select first row, second column
print(df.iloc[0, 1])         # 85

# 5. Select rows where marks are greater than or equal to 80
print(df.loc[df["marks"] >= 80])

# 6. Select the first two rows and first two columns
print(df.iloc[0:2, 0:2])

print(df.loc["A":"B"])  # Includes A and B
print(df.iloc[0:2])     # Includes positions 0 and 1
```

### Common operations

| Operation                          | Example                | Meaning                         |
| ---------------------------------- | ---------------------- | ------------------------------- |
| Select row by label                | `df.loc["A"]`          | Select row with index `A`       |
| Select row by position             | `df.iloc[0]`           | Select first row                |
| Select a specific value            | `df.loc["A", "marks"]` | Marks for index `A`             |
| Select by row and column positions | `df.iloc[0, 1]`        | First row, second column: `85`  |
| Select multiple rows               | `df.loc[["A", "C"]]`   | Rows with labels A and C        |
| Select a row range                 | `df.iloc[0:2]`         | First two rows                  |
| Select a column                    | `df.loc[:, "name"]`    | All rows from the `name` column |
| Select columns by position         | `df.iloc[:, 0]`        | All rows from the first column  |


- `loc` includes the ending label in a label-based slice.
- `iloc` excludes the ending position, following normal Python slicing.

### Quick reference

| Feature               | `loc`                      | `iloc`                          |
| --------------------- | -------------------------- | ------------------------------- |
| Selection method      | Labels                     | Integer positions               |
| Row example           | `df.loc["A"]`              | `df.iloc[0]`                    |
| Column example        | `df.loc[:, "marks"]`       | `df.iloc[:, 1]`                 |
| Conditional filtering | `df.loc[df["marks"] > 80]` | Boolean arrays can also be used |
| Slice endpoint        | Included for label slices  | Excluded for positional slices  |

Remember: Use `loc` when you know the row labels or column names. Use `iloc` when you want to select data by its position.
