```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Amar", "Rahul", "Priya"],
    "marks": [85, 72, 91],
    "age": [25, 27, 24]
}, index=["A", "B", "C"])

print("\n---------------------DataFrame:---------------------")
print(df)
# 1. Select a row using its label
print("\n---------------------Row 'B':---------------------")
print(df.loc["B"])

# 2. Select the first row using its position
print("\n---------------------First row:---------------------")
print(df.iloc[0])

# 3. Select marks for index A
print("\n---------------------Marks for index 'A':---------------------")
print(df.loc["A", "marks"])  # 85

# 4. Select first row, second column
print("\n---------------------First row, second column:---------------------")
print(df.iloc[0, 1])         # 85

# 5. Select rows where marks are greater than or equal to 80
print("\n---------------------Rows with marks >= 80:---------------------")
print(df.loc[df["marks"] >= 80])

# 6. Select the first two rows and first two columns
print("\n---------------------First two rows and columns:---------------------")
print(df.iloc[0:2, 0:2])

print("\n---------------------Rows 'A' to 'B':---------------------")
print(df.loc["A":"B"])  # Includes A and B
print("\n---------------------Positions 0 to 1:---------------------")
print(df.iloc[0:2])     # Includes positions 0 and 1
```

### Common operations

| Operation                          | Example                | Meaning                         |
| ---------------------------------- | ---------------------- | ------------------------------- |
| Select row by label                | `df.loc["A"]`          | Select row with index `A`       |
| Select row by position             | `df.iloc[0]`           | Select first row                |
| Select a specific value            | `df.loc["A", "marks"]` | Marks for index `A`             |
| Select by row and column positions | `df.iloc[0, 1]`        | First row, second column: `85`  |
| Select multiple rows               | `df.loc[["A", "C"]]`   | Rows with labels A and C        |
| Select a row range                 | `df.iloc[0:2]`         | First two rows                  |
| Select a column                    | `df.loc[:, "name"]`    | All rows from the `name` column |
| Select columns by position         | `df.iloc[:, 0]`        | All rows from the first column  |


- `loc` includes the ending label in a label-based slice.
- `iloc` excludes the ending position, following normal Python slicing.

### Quick reference

| Feature               | `loc`                      | `iloc`                          |
| --------------------- | -------------------------- | ------------------------------- |
| Selection method      | Labels                     | Integer positions               |
| Row example           | `df.loc["A"]`              | `df.iloc[0]`                    |
| Column example        | `df.loc[:, "marks"]`       | `df.iloc[:, 1]`                 |
| Conditional filtering | `df.loc[df["marks"] > 80]` | Boolean arrays can also be used |
| Slice endpoint        | Included for label slices  | Excluded for positional slices  |

Remember: Use `loc` when you know the row labels or column names. Use `iloc` when you want to select data by its position.

```python
import pandas as pd


# --------------------------------------------------
# Create sample dataset
# --------------------------------------------------

df = pd.DataFrame({
    "name": ["Amar", "Rahul", "Priya", "Ankit", "Neha"],
    "study_hours": [5, 3, 7, 4, 8],
    "attendance": [90, 75, 95, 80, 88],
    "marks": [85, 65, 92, 70, 96]
})

print("Original DataFrame:")
print(df)
```

```python
# --------------------------------------------------
# 1. Filter rows
# --------------------------------------------------

high_scorers = df[df["marks"] >= 80]

print("\nStudents with marks >= 80:")
print(high_scorers)
```

```python
# --------------------------------------------------
# 2. Multiple conditions - AND
# --------------------------------------------------

selected = df[
    (df["marks"] >= 80) &
    (df["attendance"] >= 90)
]

print("\nMarks >= 80 AND attendance >= 90:")
print(selected)

```

```python
# --------------------------------------------------
# 3. Multiple conditions - OR
# --------------------------------------------------

selected_or = df[
    (df["marks"] >= 90) |
    (df["study_hours"] >= 7)
]

print("\nMarks >= 90 OR study hours >= 7:")
print(selected_or)

```

```python
# --------------------------------------------------
# 4. isin()
# --------------------------------------------------

selected_names = df[
    df["name"].isin(["Amar", "Priya"])
]

print("\nSelected students:")
print(selected_names)

```

```python
# --------------------------------------------------
# 5. between()
# --------------------------------------------------

medium_scores = df[
    df["marks"].between(70, 90)
]

print("\nMarks between 70 and 90:")
print(medium_scores)
```

```python
# --------------------------------------------------
# 6. Sort ascending
# --------------------------------------------------

sorted_marks = df.sort_values("marks")

print("\nSorted by marks - ascending:")
print(sorted_marks)
```

```python
# --------------------------------------------------
# 7. Sort descending
# --------------------------------------------------

sorted_marks_desc = df.sort_values(
    "marks",
    ascending=False
)

print("\nSorted by marks - descending:")
print(sorted_marks_desc)
```

```python
# --------------------------------------------------
# 8. Sort by multiple columns
# --------------------------------------------------

sorted_multiple = df.sort_values(
    ["marks", "attendance"],
    ascending=[False, False]
)

print("\nSorted by marks and attendance:")
print(sorted_multiple)
```

```python
# --------------------------------------------------
# 9. Reset index
# --------------------------------------------------

result = df[df["marks"] >= 80]

result = result.reset_index(drop=True)

print("\nFiltered data with reset index:")
print(result)
```

|---|---|---|

```python
import pandas as pd
import numpy as np


# --------------------------------------------------
# Create dataset with data-quality problems
# --------------------------------------------------

df = pd.DataFrame({
    "name": ["Amar", "Rahul", "Priya", "Ankit", "Priya", "Neha"],
    "age": [25, 27, np.nan, 24, np.nan, 26],
    "marks": [85, np.nan, 92, 70, 92, 88],
    "city": ["Mumbai", "Delhi", "Bangalore", "Mumbai", "Bangalore", None]
})

print("Original DataFrame:")
print(df)
```

```python
# --------------------------------------------------
# 1. Detect missing values
# --------------------------------------------------

print("\nMissing values:")
print(df.isna())
```

```python
# --------------------------------------------------
# 2. Count missing values
# --------------------------------------------------

print("\nMissing count per column:")
print(df.isna().sum())
```

```python
# --------------------------------------------------
# 3. Remove rows containing missing values
# --------------------------------------------------

df_dropped = df.dropna()

print("\nAfter dropping rows with missing values:")
print(df_dropped)
```

```python
# --------------------------------------------------
# 4. Fill missing numeric values with mean
# --------------------------------------------------

df["age"] = df["age"].fillna(df["age"].mean())

df["marks"] = df["marks"].fillna(df["marks"].mean())

print("\nAfter filling numeric missing values:")
print(df)
```

```python
# --------------------------------------------------
# 5. Fill missing categorical values
# --------------------------------------------------

df["city"] = df["city"].fillna("Unknown")

print("\nAfter filling city:")
print(df)
```

```python
# --------------------------------------------------
# 6. Detect duplicate rows
# --------------------------------------------------

print("\nDuplicate rows:")
print(df.duplicated())
```

```python
# --------------------------------------------------
# 7. Count duplicate rows
# --------------------------------------------------

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
# --------------------------------------------------
# 8. Remove duplicate rows
# --------------------------------------------------

df = df.drop_duplicates()

print("\nAfter removing duplicates:")
print(df)
```

```python
# --------------------------------------------------
# 9. Replace incorrect/unwanted values
# --------------------------------------------------

df["city"] = df["city"].replace({
    "Bangalore": "Bengaluru"
})

print("\nAfter replacing city name:")
print(df)
```

```python
import pandas as pd
import numpy as np


# --------------------------------------------------
# Create dataset
# --------------------------------------------------

df = pd.DataFrame({
    "name": ["Amar", "Rahul", "Priya", "Ankit", "Neha"],
    "study_hours": [5, 3, 7, 4, 8],
    "attendance": [90, 75, 95, 80, 88],
    "marks": [85, 65, 92, 70, 96]
})

print("Original DataFrame:")
print(df)
```

```python
# --------------------------------------------------
# 1. Create a new column using arithmetic
# --------------------------------------------------

df["study_minutes"] = df["study_hours"] * 60

print("\nAfter creating study_minutes:")
print(df)
```

```python
# --------------------------------------------------
# 2. Create a feature using multiple columns
# --------------------------------------------------

df["performance_score"] = (
    df["marks"] * 0.8 +
    df["attendance"] * 0.2
)

print("\nPerformance score:")
print(df)
```

```python
# --------------------------------------------------
# 3. Create a categorical column using np.where()
# --------------------------------------------------

df["result"] = np.where(
    df["marks"] >= 70,
    "Pass",
    "Fail"
)

print("\nResult:")
print(df)
```

```python
# --------------------------------------------------
# 4. Create multiple categories using apply()
# --------------------------------------------------

def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"


df["grade"] = df["marks"].apply(get_grade)

print("\nGrade:")
print(df)
```

```python
# --------------------------------------------------
# 5. Map existing values
# --------------------------------------------------

grade_points = {
    "A": 4,
    "B": 3,
    "C": 2,
    "D": 1
}

df["grade_points"] = df["grade"].map(grade_points)

print("\nGrade points:")
print(df)
```

```python
# --------------------------------------------------
# 6. Rename columns
# --------------------------------------------------

df = df.rename(columns={
    "study_hours": "daily_study_hours",
    "marks": "exam_marks"
})

print("\nAfter renaming columns:")
print(df)
```

```python
# --------------------------------------------------
# 7. Drop an unnecessary column
# --------------------------------------------------

df = df.drop(columns=["grade_points"])

print("\nAfter dropping grade_points:")
print(df)
```

| Operation | Example | Purpose |
|---|---|---|
| Create column | `df["total"] = ...` | Add a derived column |
| Arithmetic | `df["score"] * 2` | Transform numeric data |
| Conditional column | `np.where(...)` | Create values based on condition |
| `apply()` | `df["marks"].apply(...)` | Apply a Python function |
| `map()` | `df["grade"].map(...)` | Map existing values to new values |
| Rename column | `df.rename(...)` | Change column names |
| Drop column | `df.drop(...)` | Remove unnecessary columns |

| Situation | Prefer |
|---|---|
| `marks * 2` | Vectorized Pandas/NumPy |
| `marks >= 70` | Boolean condition / `np.where()` |
| `A → 1, B → 2` | `map()` |
| Complex custom Python logic | `apply()` |
| Function needs entire row | `apply(..., axis=1)` |

For your ML preparation, the important thing is not memorizing apply() syntax. Understand that it is a way to pass each Series value (or each DataFrame row/column) through your own function.

```python
import pandas as pd


df = pd.DataFrame({
    "name": ["Amar", "Rahul", "Priya", "Ankit", "Neha", "Ravi"],
    "city": [
        "Bengaluru",
        "Delhi",
        "Bengaluru",
        "Delhi",
        "Bengaluru",
        "Delhi"
    ],
    "marks": [85, 65, 92, 70, 96, 80],
    "study_hours": [5, 3, 7, 4, 8, 6]
})
print("Original DataFrame:")
print(df)
```

```python
# 1. Average marks by city
avg_marks = df.groupby("city")["marks"].mean()

print("\nAverage marks:")
print(avg_marks)
```

```python
# 2. Maximum marks by city
max_marks = df.groupby("city")["marks"].max()

print("\nMaximum marks:")
print(max_marks)
```

```python
# 3. Number of students by city
student_count = df.groupby("city").size()

print("\nStudent count:")
print(student_count)
```

```python
# 4. Multiple statistics
summary = df.groupby("city")["marks"].agg(
    ["mean", "min", "max", "count"]
)

print("\nMarks summary:")
print(summary)
```

```python
# 5. Multiple columns
summary = df.groupby("city").agg({
    "marks": ["mean", "max"],
    "study_hours": ["mean", "sum"]
})

print("\nComplete summary:")
print(summary)
```

| `count()` | Number of values |
| `min()` | Minimum |
| `max()` | Maximum |
| `median()` | Median |

```python
import pandas as pd


# --------------------------------------------------
# Customer Data
# --------------------------------------------------

customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104],
    "name": ["Amar", "Rahul", "Priya", "Neha"],
    "city": ["Bengaluru", "Delhi", "Mumbai", "Pune"]
})


# --------------------------------------------------
# Transaction Data
# --------------------------------------------------

transactions = pd.DataFrame({
    "customer_id": [101, 102, 101, 105],
    "amount": [5000, 3000, 7000, 4000]
})

print("Customer Data:")
print(customers)
print("\nTransaction Data:")
print(transactions)
```

```python
# --------------------------------------------------
# 1. Inner merge
# --------------------------------------------------

inner = pd.merge(
    customers,
    transactions,
    on="customer_id",
    how="inner"
)

print("\nINNER JOIN:")
print(inner)
```

```python
# --------------------------------------------------
# 2. Left merge
# --------------------------------------------------

left = pd.merge(
    customers,
    transactions,
    on="customer_id",
    how="left"
)
print("\nLEFT JOIN:")
print(left)
```

```python
# --------------------------------------------------
# 3. Outer merge
# --------------------------------------------------

outer = pd.merge(
    customers,
    transactions,
    on="customer_id",
    how="outer"
)

print("\nOUTER JOIN:")
print(outer)
```

```python
# --------------------------------------------------
# 4. Concatenate rows
# --------------------------------------------------

jan = pd.DataFrame({
    "customer_id": [101, 102],
    "amount": [5000, 3000]
})

feb = pd.DataFrame({
    "customer_id": [103, 104],
    "amount": [7000, 4000]
})

combined = pd.concat(
    [jan, feb],
    ignore_index=True
)

print("\nCONCAT:")
print(combined)
```

```python
customers = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "name": ["Amar", "Rahul", "Priya"]
})

transactions = pd.DataFrame({
    "cust_id": [101, 102, 103],
    "amount": [5000, 3000, 7000]
})

result = pd.merge(
    customers,
    transactions,
    left_on="customer_id",
    right_on="cust_id",
    how="inner"
)

print(result)
```

| Operation | Think |
|---|---|
| `merge()` | **Join using a key** |
| `join()` | **Join using index** |
| `concat()` | **Stack DataFrames** |
| `inner` | Matching records |
| `left` | Keep left |
| `right` | Keep right |
| `outer` | Keep everything |





| Situation | Use |
|---|---|
| Same key name in both DataFrames | `on="customer_id"` |
| Different key names | `left_on="customer_id", right_on="cust_id"` |
| Different key names + want all customers | Add `how="left"` |
